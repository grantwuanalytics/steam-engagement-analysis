```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
from pathlib import Path

project_dir = Path.cwd()
data_dir = project_dir.parent / "Data"

df_games = pd.read_csv(data_dir / "steam_games_cleaned.csv")
df_price_tiers = pd.read_csv(data_dir / "price_tier_summary.csv")
df_genres = pd.read_csv(data_dir / "genre_summary.csv")

print(f"Games: {df_games.shape}")
print(f"Price tiers: {df_price_tiers.shape}")
print(f"Genres: {df_genres.shape}")
```

```python
total_games = len(df_games)
total_reviews = df_games["total_reviews"].sum()

print(f"Total Games: {total_games:,}")
print(f"Total Reviews: {total_reviews:,}")
```

```python
metrics = ["price", "positive_pct", "avg_hours_played", "peak_ccu"]

# Examine distributions and relationships between key metrics
print("Metric Skewness")
print(df_games[metrics].skew())

print("\nPearson Correlation")
display(df_games[metrics].corr(method="pearson"))

print("\nSpearman Correlation")
display(df_games[metrics].corr(method="spearman"))
```

```python
# Correlation heatmaps
labels = {
    "price": "Price ($)",
    "positive_pct": "Positive Reviews (%)",
    "avg_hours_played": "Avg. Hours Played",
    "peak_ccu": "Peak Concurrent Players"
}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, method, title in zip(
    axes,
    ["pearson", "spearman"],
    ["Pearson Correlation", "Spearman Correlation"]
):
    correlation = df_games[metrics].corr(method=method)
    correlation = correlation.rename(index=labels, columns=labels)

    sns.heatmap(
        correlation,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        ax=ax,
        cbar=(method == "spearman")
    )
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.tick_params(axis="x", rotation=45)
    ax.tick_params(axis="y", rotation=0)

fig.suptitle(
    "Pearson vs. Spearman: Relationships in Steam Game Metrics",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.savefig("steam_correlation_heatmaps.png", dpi=300, bbox_inches="tight")
plt.show()


# Price vs. player engagement
df_scatter = df_games[
    (df_games["price"] > 0) &
    (df_games["avg_hours_played"] > 0)
].copy()

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_scatter,
    x="price",
    y="avg_hours_played",
    alpha=0.15,
    s=20,
    edgecolor=None
)

plt.yscale("log")
plt.xlim(0, 70)
plt.title("Game Price vs. Average Hours Played", fontsize=13, fontweight="bold")
plt.xlabel("Price ($ USD)")
plt.ylabel("Average Hours Played (Log Scale)")

plt.tight_layout()
plt.savefig("steam_price_engagement.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Prepare genre data for visualization
valid_genres = [
    "Action", "Adventure", "Casual", "Early Access",
    "Indie", "Massively Multiplayer", "Racing", "RPG",
    "Simulation", "Sports", "Strategy"
]

df_genres_clean = df_genres[
    df_genres["clean_genre"].isin(valid_genres)
].copy()

df_genres_clean = df_genres_clean.sort_values(
    "retention_ratio", ascending=False
)

print(f"Genres included: {len(df_genres_clean)}")
display(df_genres_clean)
```

```python
# Genre performance visualization
fig, ax = plt.subplots(figsize=(12, 7))

scatter = ax.scatter(
    df_genres_clean["retention_ratio"],
    df_genres_clean["avg_positive_pct"],
    s=df_genres_clean["total_games"] / 12,
    c=df_genres_clean["retention_ratio"],
    cmap="viridis",
    alpha=0.85,
    edgecolors="white",
    linewidths=1
)

# Position labels below each bubble based on bubble size
for _, row in df_genres_clean.iterrows():
    bubble_size = row["total_games"] / 12
    offset = -(np.sqrt(bubble_size / np.pi) + 8)

    ax.annotate(
        row["clean_genre"],
        (row["retention_ratio"], row["avg_positive_pct"]),
        xytext=(0, offset),
        textcoords="offset points",
        ha="center",
        va="top",
        fontsize=9
    )

# Reference lines based on median genre performance
median_retention = df_genres_clean["retention_ratio"].median()
median_sentiment = df_genres_clean["avg_positive_pct"].median()

ax.axvline(median_retention, color="gray", linestyle="--", alpha=0.6)
ax.axhline(median_sentiment, color="gray", linestyle="--", alpha=0.6)

ax.set_title(
    "Steam Genre Performance: Engagement vs. Player Sentiment",
    fontsize=15,
    fontweight="bold",
    pad=15
)
ax.set_xlabel("Retention Ratio (Avg. Hours Played / Peak CCU)")
ax.set_ylabel("Average Positive Reviews (%)")

ax.set_xlim(45, 215)
ax.set_ylim(64, 85)
ax.grid(alpha=0.25)
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig("steam_genre_performance.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
from pathlib import Path

output_dir = Path.cwd() / "exports"
output_dir.mkdir(parents=True, exist_ok=True)

df_games.to_csv(output_dir / "steam_games_processed.csv", index=False)
df_price_tiers.to_csv(output_dir / "price_tier_summary.csv", index=False)
df_genres_clean.to_csv(output_dir / "genre_summary.csv", index=False)

print("Datasets exported successfully.")
```
